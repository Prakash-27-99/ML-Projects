```python
import pandas as pd

# Load the Excel files
inventory = pd.read_csv("inventory.csv")
products = pd.read_csv("products.csv")
sales = pd.read_csv("sales.csv")
stores = pd.read_csv("stores.csv")

# Merge sales with products (on Product_ID)
merged_df = sales.merge(products, on="Product_ID", how="left")

# Merge with stores (on Store_ID)
merged_df = merged_df.merge(stores, on="Store_ID", how="left")

# Merge with inventory (on Store_ID & Product_ID)
merged_df = merged_df.merge(inventory, on=["Store_ID", "Product_ID"], how="left")

# Drop duplicate or unnecessary columns (if any)
merged_df = merged_df.drop_duplicates()

# Save the cleaned merged dataset
merged_df.to_csv("final_combined_data.csv", index=False)

print("Merged and cleaned dataset saved as 'final_combined_data.csv'.")
```
